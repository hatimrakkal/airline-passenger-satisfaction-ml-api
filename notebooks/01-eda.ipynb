```python
airline_data.dropna(inplace=True)
print(airline_data.isna().sum())
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(x="satisfaction", y="Age", data=airline_data, hue="Gender")

plt.title("Age distribution by satisfaction and Gender")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,4))

sns.boxplot(x="satisfaction", y="Flight Distance", data=airline_data)
plt.title("Flight Distance distribution by satisfaction")
plt.show()
```

* satisfied passengers appear to have a longer flight distance.
* in the neutral or dissatisfied group, there are many outliers.
* therefore the relationship between distance and satisfaction is not necessarily direct.

```python
plt.figure(figsize=(12,4))

sns.boxplot(x="Class", y="Flight Distance", data=airline_data)
plt.title("Distribution de la distance selon Class")
plt.show()
```

- The boxplot shows that satisfied passengers appear to take longer flights.
- A plausible hypothesis is that long trips are more often associated with Business class or business travel.
- In this case, the observed satisfaction would not be due solely to distance, but also to other variables such as class and type of travel.
- Thus, flight distance appears to be associated with satisfaction, but this relationship may be influenced by other factors.

```python
fig, axes = plt.subplots(7, 2, figsize=(30,70))
list_of_columns = airline_data.columns
count = 0
for i in range(7):
    for j in range(2):
        sns.countplot(ax=axes[i,j], data=airline_data, x=list_of_columns[6+count], hue="satisfaction", hue_order=["satisfied", "neutral or dissatisfied"])
        count+=1

plt.tight_layout()
plt.show()
```

```python
sns.catplot(
    data=airline_data,
    x="Departure/Arrival time convenient",
    y="Departure Delay in Minutes",
    kind="bar"
)

plt.show()
```

```python
from src.data import NUM_VAR

# 1. Calculate the correlation matrix (only for numeric columns)
corr_matrix = airline_data[NUM_VAR].corr()

# 2. Set up the figure size
plt.figure(figsize=(16, 12))

# 3. Plot the heatmap
sns.heatmap(
    corr_matrix, 
    annot=True,          # Show the numbers inside the squares
    fmt=".2f",           # Format numbers to 2 decimal places
    cmap="coolwarm",     # Color scheme (blue = negative, red = positive)
    vmin=-1, vmax=1,     # Set the scale from -1 to 1
    linewidths=0.5       # Add lines between squares for readability
)

plt.title("Correlation Matrix of Numeric Features")
plt.show()
```
